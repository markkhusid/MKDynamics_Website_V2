## Euclid's algorithm for two integers

The GCD of $a$ and $b$ can be found by replacing $(a, b)$ with $(b, a \bmod b)$ until the remainder is zero. The last non-zero remainder (stored in $b$) is $\gcd(a,b)$.

```python
def gcd(a, b):
    """Return gcd(a, b) using Euclid's algorithm."""
    print(f"gcd({a}, {b}) ...")
    mod = a % b
    print(f"  initial remainder a % b = {mod}")

    while mod != 0:
        a = b
        b = mod
        int_div = a // b
        mod = a % b
        print(f"  a={a}, b={b}, a//b={int_div}, a%b={mod}")

    print(f"  -> gcd = {b}")
    return b

# Quick unit checks
for pair in [(48, 18), (12, 2), (15, 10), (90, 25)]:
    print(f"GCD{pair} = {gcd(*pair)}")
    print()
```

## Parse a list of integers (assignment-style input format)

The assignment accepts a string that looks like a Python list, e.g. `[24, 60, 120, 30]`.
We strip brackets, split on commas, and convert each token to `int`.

```python
# Hard-coded demo input (was: input("Input a list of integers: "))
input_str = "[24, 60, 120, 30]"
print("Input string:", input_str)
print("Type of input string:", type(input_str))

stripped_str = input_str.strip("[]")
print("Stripped string:", stripped_str)

split_stripped_str_list_1 = stripped_str.split(",")
print("After split on commas:", split_stripped_str_list_1)

split_stripped_str_list_2 = [item.strip() for item in split_stripped_str_list_1]
print("After strip of each token:", split_stripped_str_list_2)

int_list = [int(item) for item in split_stripped_str_list_2]
print("As integers:", int_list)
print("Type of first element:", type(int_list[0]))
```

## Extend pairwise GCD across the whole list

Start with $\gcd(x_0, x_1)$, then fold the rest:  
$\mathrm{current} \leftarrow \gcd(\mathrm{current}, x_i)$ for $i = 2, \ldots, n-1$.

```python
working_list = int_list.copy()
len_list = len(working_list)
print("Input list:", working_list)
print("Length of input list:", len_list)

current_gcd = gcd(working_list[0], working_list[1])
print(f"GCD of {working_list[0]} and {working_list[1]} is: {current_gcd}")
print()

for i in range(2, len_list):
    print(f"i = {i}, next value = {working_list[i]}")
    print(f"GCD of {current_gcd} and {working_list[i]} is: ", end="")
    current_gcd = gcd(current_gcd, working_list[i])
    print(current_gcd)
    print()

print("The GCD is: ", current_gcd)
```

## Takeaways

- Euclid's algorithm only needs remainder operations — no factorization.
- Folding pairwise GCDs yields the GCD of an entire list.
