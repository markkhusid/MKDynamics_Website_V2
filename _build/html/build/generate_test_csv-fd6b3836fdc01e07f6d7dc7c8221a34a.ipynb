## Random test matrix

Draws a 10 by 24 matrix of integers from 0 through 300, the same shape as a short glucose window, and writes it to `test.csv` with no row index and no header.


```python
import numpy as np
import pandas as pd

# Step 1: Generate a 10x24 matrix of random numbers between 0 and 300
matrix = np.random.randint(0, 301, size=(10, 24))

# Step 2: Convert the matrix to a pandas DataFrame
df = pd.DataFrame(matrix)

# Step 3: Save the DataFrame to a CSV file
df.to_csv('test.csv', index=False, header=False)

print("Matrix saved to 'test.csv'")
```
