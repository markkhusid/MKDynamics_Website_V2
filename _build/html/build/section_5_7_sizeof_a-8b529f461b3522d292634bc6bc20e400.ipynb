## Overview

`sizeof` on a pointer yields the size of the address, while `sizeof *p` yields the size of the pointed-to type.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    int *p;
    int x = 0;
    p = &x;

    printf("sizeof(int)  = %zu\n", sizeof(int));
    printf("sizeof p     = %zu  (size of the pointer itself)\n", sizeof p);
    printf("sizeof *p    = %zu  (size of the pointed-to int)\n", sizeof *p);
    printf("sizeof(int*) = %zu\n", sizeof(int *));
}
```


## Step-by-step explanation

1. Point `p` at an `int`.
2. Print `sizeof(int)`, `sizeof p`, and `sizeof *p`.
3. Observe that pointer size is platform-dependent (often 8 on 64-bit hosts).
4. Use `sizeof *p` when you want the pointee size without naming the type.
