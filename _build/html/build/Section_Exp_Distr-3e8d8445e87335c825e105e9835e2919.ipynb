## Overview

Generates 1000 approximately normal random values using a polar Box–Muller-style method with a saved spare deviate.


## Step-by-step explanation

1. get_exp_rand implements a polar Gaussian method (despite the filename).
2. Two uniforms are mapped to [-1,1]^2 and rejected outside the unit disk.
3. The Box–Muller factor scales the coordinates.
4. One sample is returned and the other stored for the next call.
5. The driver prints 1000 values for statistical checks or plotting.
