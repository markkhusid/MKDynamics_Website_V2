Target integral (known closed form):
$$
I_2 = \int_0^{\infty} \frac{1}{1+x^2}\,dx = \frac{\pi}{2}.
$$
We compare two numerical strategies with SciPy `quad`.


## Analytic setup: change of variables

Let $x = \tan\theta$, so $dx = \sec^2\theta\,d\theta$.
Limits map as $x=0\to\theta=0$ and $x\to\infty\to\theta=\pi/2$.
Using $1+\tan^2\theta = \sec^2\theta$, the integrand collapses to $1$:
$$
I_2 = \int_0^{\pi/2} 1\,d\theta = \frac{\pi}{2}.
$$


```python
import numpy as np
from scipy.integrate import quad

KNOWN_RESULT = np.pi / 2
PRINT_PRECISION = 12
DIFF_PRECISION = 16

print(f"Known result π/2 = {KNOWN_RESULT:.{PRINT_PRECISION}f}")
```

## Method 1 — `quad` after substitution to a finite interval

Integrate the constant integrand $1$ from $\arctan(0)$ to $\arctan(\infty) = \pi/2$.


```python
ORIG_LOWER_LIMIT = 0
ORIG_UPPER_LIMIT = np.inf

new_lower_limit = np.arctan(ORIG_LOWER_LIMIT)  # 0
new_upper_limit = np.arctan(ORIG_UPPER_LIMIT)  # π/2

def new_integrand(theta):
    """Transformed integrand: sec^2(θ)/sec^2(θ) = 1."""
    return 1

result_method_1, error_method_1 = quad(new_integrand, new_lower_limit, new_upper_limit)
diff_method_1 = abs(result_method_1 - KNOWN_RESULT)

print(f"θ limits: [{new_lower_limit}, {new_upper_limit}]")
print(f"I2 Method 1 (substituted): {result_method_1: .{PRINT_PRECISION}f}")
print(f"quad error estimate: {error_method_1:.3e}")
print(f"|I2 - π/2| Method 1: {diff_method_1: .{DIFF_PRECISION}f}")
```

## Method 2 — `quad` on the original infinite interval

SciPy can integrate improper integrals directly when the upper limit is `np.inf`.


```python
def original_integrand(x):
    """I2(x) = 1 / (1 + x^2)."""
    return 1 / (1 + x**2)

result_method_2, error_method_2 = quad(original_integrand, ORIG_LOWER_LIMIT, ORIG_UPPER_LIMIT)
diff_method_2 = abs(result_method_2 - KNOWN_RESULT)

print(f"I2 Method 2 (infinite limits): {result_method_2: .{PRINT_PRECISION}f}")
print(f"quad error estimate: {error_method_2:.3e}")
print(f"|I2 - π/2| Method 2: {diff_method_2: .{DIFF_PRECISION}f}")
```

## Takeaways

- A well-chosen substitution turns an improper integral into a trivial finite one.
- Modern `quad` also handles $\infty$ limits directly; both routes recover $\pi/2$ to high precision.
