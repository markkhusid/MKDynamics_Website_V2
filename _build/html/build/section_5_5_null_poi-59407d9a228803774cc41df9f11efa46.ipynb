## Overview

A null pointer is a special pointer value that means “points nowhere.” Always check before dereferencing.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    int *p = NULL;   /* safe default: points nowhere */

    printf("p is %p\n", (void *)p);

    if (p == NULL) {
        printf("p is NULL — not dereferencing it.\n");
    }

    /* Uncommenting the next line is undefined behavior / likely crash:
     * *p = 12;
     */

    int x = 7;
    p = &x;
    if (p != NULL) {
        printf("p now points to x; *p = %d\n", *p);
    }
}
```


## Step-by-step explanation

1. Initialize `p` to `NULL` and print it.
2. Branch on `p == NULL` so we do not dereference an empty pointer.
3. Point `p` at a real variable and safely print `*p`.
4. Note that writing through a null pointer is undefined behavior.
