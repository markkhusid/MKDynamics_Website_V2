                The Esper cipher in this course is a byte cipher with two steps. Each byte is rotated left by a small bit count, then exclusive-ored with a byte from a repeating key. Decryption exclusive-ors the same key byte and rotates right by the same count. Both operations undo themselves when they are applied in that reverse order.

                The submission decrypted a course ciphertext with a course key. This notebook uses a new seven-byte key and a new sentence. The course key and ciphertext are not on the page.
                


## Rotate, then exclusive-or


```python
def rotl(byte, count):
    count &= 7
    byte &= 0xFF
    if count == 0:
        return byte
    return ((byte << count) | (byte >> (8 - count))) & 0xFF


def rotr(byte, count):
    count &= 7
    if count == 0:
        return byte & 0xFF
    return rotl(byte, 8 - count)


def encrypt(data, key, count):
    return bytes(
        rotl(byte, count) ^ key[index % len(key)]
        for index, byte in enumerate(data)
    )


def decrypt(data, key, count):
    return bytes(
        rotr(byte ^ key[index % len(key)], count)
        for index, byte in enumerate(data)
    )


key = b"Bright7"
count = 3
plain = b"Public demo only"
cipher = encrypt(plain, key, count)
recovered = decrypt(cipher, key, count)
print(cipher.hex())
print(recovered)
assert recovered == plain
```

                ## Why the order matters

                Exclusive-or is its own inverse, and a right rotation undoes a left rotation. Reversing the order, or using a different count, leaves the bytes scrambled. The three lines below show a matching decrypt, a swapped order, and the wrong count.
                


```python
swapped = bytes(
    rotr(byte, count) ^ key[index % len(key)]
    for index, byte in enumerate(cipher)
)
wrong_count = decrypt(cipher, key, count + 1)
print("matching ", recovered)
print("swapped  ", swapped)
print("wrong    ", wrong_count)
assert swapped != plain
assert wrong_count != plain
```
