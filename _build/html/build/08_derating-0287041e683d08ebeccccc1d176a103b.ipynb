```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Directory that contains opamp_design/. Works when this notebook is run
# from the standalone repo (repo root or notebooks/) or from this website
# section (the folder that holds these notebooks and opamp_design/).
def _find_project_root() -> Path:
    start = Path.cwd().resolve()
    named = Path("Sections/Electronics/Analog_and_Digital/OPA365_Fig86")
    candidates = [start, *start.parents, start / named]
    if start.name == "notebooks":
        candidates.insert(0, start.parent)
    for cand in candidates:
        if (cand / "opamp_design" / "__init__.py").is_file():
            return cand
    return start

ROOT = _find_project_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from opamp_design.params import CircuitParams, OpampParams, ToleranceSpec, format_eng
from opamp_design import analysis, schematic

# -----------------------------------------------------------------------------
# EDITABLE PARAMETERS — change R/C/tolerances here for what-if studies.
# For a completely different netlist, keep these as documentation of the
# stimulus/supply and drive opamp_design.simulate.run_external_netlist(...)
# with your own .cir instead of the Fig. 8-6 generators.
# -----------------------------------------------------------------------------
R1 = 1.8e3       # ohms
R2 = 19.5e3
R3 = 150e3
C1 = 3.3e-9      # farads
C2 = 47e-12
C3 = 220e-12
R_TOL = 0.001    # 0.1% as fraction
C_TOL = 0.01     # 1% as fraction
V_SUPPLY = 5.0
V_IN_BIAS = 2.5
V_IN_PEAK = 2.5
F_DESIGN = 20e3

p = CircuitParams(
    R1=R1, R2=R2, R3=R3, C1=C1, C2=C2, C3=C3,
    r_tol=ToleranceSpec(R_TOL),
    c_tol=ToleranceSpec(C_TOL),
)
# Operating point overrides
from dataclasses import replace
p = replace(p, op=replace(p.op, v_supply=V_SUPPLY, v_in_bias=V_IN_BIAS,
                          v_in_peak=V_IN_PEAK, f_design_hz=F_DESIGN))

FIG = ROOT / "results" / "figures"
FIG.mkdir(parents=True, exist_ok=True)
print("ROOT =", ROOT)
print("R1..C3 =", p.passive_dict())
```

```python
from opamp_design.derating import estimate_stresses, evaluate_derating, power_derating_curve

stresses = estimate_stresses(p)
for s in stresses:
    print(f"{s.name:8s}  Vpk={s.v_peak:.4g} V  I~{s.i_rms_est:.3g} A  P~{s.p_est:.3g} W  ({s.notes})")

table = evaluate_derating(p, temp_c=25.0, stresses=stresses)
display(table)

# Save
out = ROOT / "results" / "derating_table_25C.csv"
table.to_csv(out, index=False)
print("Wrote", out)
```

## Passband stress versus rating and derating

The stresses above are the DC bias solution, which puts almost no signal current in the resistors. This section applies the same 10 kHz, 1 V peak sine about 2.5 V used in the transient notebooks.

Resistor dissipation is the average of that waveform, \(V_{\mathrm{dc}}^2/R + |V_{\mathrm{ac}}|^2/(2R)\). Capacitor voltage is the peak, \(|V_{\mathrm{dc}}| + |V_{\mathrm{ac}}|\). The op-amp supply is the 5 V rail. Its dissipation is \(V_s \times 4.6\,\mathrm{mA}\) (the datasheet typical quiescent current) plus the small load power in the 10 kΩ resistor.

Limits are the ones already used in this study. Resistors: 0.125 W package, full rating to 70 °C, and a 50 % use of that derated power; the working-voltage guideline is 150 V. Capacitors: 50 V rating, recommended operation at half of that. The op-amp recommended operating supply used here is 5.5 V, and the package allowance used here is 0.5 W.


```python
from opamp_design.derating import passband_stresses, stress_vs_limits

stress = passband_stresses(p, freq_hz=10e3, amp_v=1.0)
lim = stress_vs_limits(p, temp_c=25.0, stresses=stress)

def _grouped(ax, labels, applied, derated, rated, ylabel, title, log=False, legend=True):
    x = np.arange(len(labels))
    w = 0.26
    b0 = ax.bar(x - w, applied, w, label="Applied", color="C0", edgecolor="k", linewidth=0.4)
    ax.bar(x, derated, w, label="Derated limit", color="C1", edgecolor="k", linewidth=0.4)
    ax.bar(x + w, rated, w, label="Rated limit", color="0.75", edgecolor="k", linewidth=0.4)
    ax.set_xticks(x)
    ax.set_xticklabels(labels)
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    ax.grid(True, axis="y", alpha=0.3)
    if legend:
        ax.legend(fontsize=8, loc="upper left")
    if log:
        ax.set_yscale("log")
    top = float(max(np.max(rated), np.max(applied), np.max(derated)))
    for rect, val in zip(b0, applied):
        color = "k"
        if log:
            y = val * 1.45
            va = "bottom"
        elif val > 0.72 * top:
            y = val * 0.55
            va = "center"
            color = "white"
        else:
            y = val + 0.025 * top
            va = "bottom"
        if val < 1e-3:
            text = f"{val * 1e6:.1f} µW"
        elif ylabel.startswith("Power"):
            text = f"{val * 1e3:.2f} mW"
        else:
            text = f"{val:.2f} V"
        ax.text(
            rect.get_x() + rect.get_width() / 2,
            y,
            text,
            ha="center",
            va=va,
            fontsize=8,
            color=color,
        )

volt = lim[lim["quantity"] == "voltage"].copy()
# Resistor peaks are a few tenths of a volt against a 150 V guideline.
# They stay in the table. Capacitors and the op-amp supply are plotted on
# separate axes so a 50 V rating does not hide the 5.5 V rail.
caps = volt[volt["component"].isin(["C1", "C2", "C3"])]
amp = volt[volt["component"] == "OPA365"]
power = lim[lim["quantity"] == "power"].copy()

fig, axes = plt.subplots(1, 2, figsize=(10.5, 4.4))
_grouped(
    axes[0],
    caps["component"].tolist(),
    caps["applied"].to_numpy(),
    caps["derated_limit"].to_numpy(),
    caps["rated_limit"].to_numpy(),
    "Voltage (V)",
    "Capacitor peak voltage",
    legend=False,
)
_grouped(
    axes[1],
    amp["component"].tolist(),
    amp["applied"].to_numpy(),
    amp["derated_limit"].to_numpy(),
    amp["rated_limit"].to_numpy(),
    "Voltage (V)",
    "Op-amp supply",
    legend=False,
)
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper center", ncol=3, frameon=False, bbox_to_anchor=(0.5, 1.08))
fig.tight_layout()
fig.savefig(FIG / "08_passband_voltage_limits.png", dpi=150, bbox_inches="tight")
plt.show()

fig, ax = plt.subplots(figsize=(8.5, 4.6))
_grouped(
    ax,
    power["component"].tolist(),
    power["applied"].to_numpy(),
    power["derated_limit"].to_numpy(),
    power["rated_limit"].to_numpy(),
    "Power (W)",
    "Average dissipation at 10 kHz vs rating and derating",
    log=True,
)
fig.tight_layout()
fig.savefig(FIG / "08_passband_power_limits.png", dpi=150)
plt.show()

rows = []
for _, r in lim.iterrows():
    if r["quantity"] == "voltage":
        applied, derated, rated, unit = r["applied"], r["derated_limit"], r["rated_limit"], "V"
        fmt = "{:.3f}"
    else:
        applied, derated, rated, unit = r["applied"] * 1e3, r["derated_limit"] * 1e3, r["rated_limit"] * 1e3, "mW"
        fmt = "{:.4f}"
    rows.append(
        {
            "component": r["component"],
            "quantity": f"{r['quantity']} ({unit})",
            "applied": fmt.format(applied),
            "derated_limit": fmt.format(derated),
            "rated_limit": fmt.format(rated),
            "utilization_of_derated_%": f"{100 * r['utilization_of_derated']:.4f}",
            "notes": r["notes"],
        }
    )
display(pd.DataFrame(rows))
if bool((lim["utilization_of_derated"] < 1.0).all()):
    print("Every applied stress is below its derated limit.")
else:
    print("At least one applied stress meets or exceeds its derated limit.")
```

### Reading the bars

C1 and C2 peak near 3.4 V. C3 peaks near 0.41 V. The recommended capacitor ceiling is 25 V and the rating is 50 V, so the voltage margin is large. The op-amp rail is 5.0 V against the 5.5 V recommended operating maximum used in this study, about 91 % of that ceiling. That is the closest of these voltage comparisons, which is why it has its own axis.

Resistor peaks are a few tenths of a volt (R1 0.36 V, R2 0.12 V, R3 0.41 V). They are listed in the table against the 150 V working-voltage guideline.

Dissipation is 36 µW in R1 and under 1 µW in R2 and R3. The derated allowance of an 0805 at 25 °C, after the 50 % use fraction, is 62.5 mW, and the package rating is 125 mW. The log axis is what makes those microwatt bars visible next to the allowance. Op-amp dissipation is 23 mW against the 0.5 W package allowance used here. Every row in the table is under its derated limit.
