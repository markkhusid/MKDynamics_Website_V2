## Overview

Shows precision loss and truncation when converting between larger and smaller kinds.


## Step-by-step explanation

1. A huge double (1e100) is converted to REAL32.
2. If the value exceeds single-precision range, the result may become infinity.
3. A 64-bit integer 2**40 is converted to INT32, which cannot represent it fully.
4. The printed INT32 result demonstrates silent truncation on narrowing conversion.
5. Convert consciously when kinds differ in range or precision.
