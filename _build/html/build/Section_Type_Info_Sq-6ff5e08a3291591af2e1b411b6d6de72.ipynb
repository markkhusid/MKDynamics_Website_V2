## Overview

Compares exact equality vs tolerance checks for sqrt(2)**2 in single and double precision.


## Step-by-step explanation

1. Compute sqrt(2) in REAL32 and REAL64.
2. Test whether squaring recovers exactly 2.0 (usually false in binary floating point).
3. Retest with absolute tolerances appropriate to each precision.
4. Scientific code should compare reals with tolerances, not ==.
5. Double precision typically passes a tighter tolerance than single precision.
