                A Caesar cipher shifts each letter by a fixed amount and wraps at the end of the alphabet. There are 25 nontrivial shifts of the English capitals. With the shift known, decryption is the same walk in the other direction.

                The course submission read a one-letter shift and a ciphertext file. This notebook does not use that file. It encrypts and decrypts a short original sentence so the round trip is visible on the page.
                


## Shift one letter


```python
def shift_char(ch, amount):
    if "A" <= ch <= "Z":
        return chr((ord(ch) - ord("A") + amount) % 26 + ord("A"))
    return ch


def caesar(text, amount):
    return "".join(shift_char(ch, amount) for ch in text)


plain = "MEET AT THE NORTH GATE"
shift = 3
cipher = caesar(plain, shift)
recovered = caesar(cipher, -shift)
print(cipher)
print(recovered)
assert recovered == plain
```

                ## Every other shift

                The same sentence under each of the other 25 shifts. Only one of them is the plaintext. That is the whole search when the shift is unknown and the text is known to be capitals.
                


```python
for amount in range(1, 26):
    print(f"{amount:2d}  {caesar(plain, amount)}")
```
