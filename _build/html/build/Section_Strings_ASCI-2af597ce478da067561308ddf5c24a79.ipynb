## Overview

Prints characters A–Z with their ASCII codes using iachar and achar.


## Step-by-step explanation

1. Loop index i runs from iachar("A") to iachar("Z").
2. achar(i) converts a code back to a character.
3. Each line prints the character and its integer code.
4. Useful for understanding collating sequences and text processing.
