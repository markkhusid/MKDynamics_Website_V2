## Overview

Shows Fortran optional arguments by clamping sample values with and without an explicit lower bound.


## Step-by-step explanation

1. An array of sample values is defined, including negatives, zero, and large positives.
2. For each value, clamp is called twice: once with only max_val, once with both min_val and max_val.
3. Inside clamp, present(min_val) detects whether the optional lower bound was supplied.
4. Values outside the effective bounds are overwritten with the nearest bound.
5. The program prints the original value and both clamped results for comparison.
