## Overview

Lists prime numbers up to 100 using a simple trial-division module function is_prime.


## Step-by-step explanation

1. primes_mod exposes is_prime(n).
2. Numbers <= 1 are not prime.
3. For n > 1, test divisors from 2 through sqrt(n).
4. The driver loops n = 0..100 and prints primes.
5. Modules separate reusable logic (src) from the program driver (app).
