## Overview

Shows how to query, get, and put the Fortran RNG seed, and that reseeding reproduces a sequence.


## Step-by-step explanation

1. random_seed(size=...) reports how many integers the seed needs.
2. Allocate seed arrays and random_seed(get=...) to capture the current seed.
3. Draw a few random_number samples and print them.
4. Put the saved seed back with random_seed(put=...).
5. Drawing again reproduces the same sequence after reseeding.
