## Optimized trial division: `is_prime(n)`

- $n \le 1$: not prime  
- $n \in \{2,3\}$: prime  
- divisible by 2 or 3: not prime  
- else test divisors of the form $i, i+2$ for $i = 5, 11, 17, \ldots$ while $i^2 \le n$

```python
def is_prime(n):
    """Return True if n is prime (optimized trial division)."""
    if n <= 1:
        print(f"  Not prime because n <= 1: {n}")
        return False
    elif (n == 2) or (n == 3):
        print(f"  Prime because n == 2 or n == 3: {n}")
        return True
    elif (n % 2 == 0) or (n % 3 == 0):
        print(f"  Not prime because n % 2 == 0 or n % 3 == 0: {n}")
        return False

    i = 5
    while i * i <= n:
        if (n % i == 0) or (n % (i + 2) == 0):
            print(f"  Not prime because n % {i} == 0 or n % {i + 2} == 0: {n}")
            return False
        i += 6

    print(f"  {n} is a prime")
    return True

# Quick smoke tests
for n in [100, 101, 2, 3, 4, 5, 99]:
    print(f"is_prime({n}) -> {is_prime(n)}")
    print()
```

## Parse mixed numeric tokens

The demo input includes integers written as floats (`31.0`) and a true non-integer (`81.4`).
We convert each token with `float`, then use `.is_integer()` before casting to `int`.

```python
# Hard-coded demo (was: input("Input a list of integers: "))
input_str = "[31.0, 17, 81.4, 28]"
print("Input string:", input_str)
print("Type of input string:", type(input_str))

stripped_str = input_str.strip("[]")
print("Stripped string:", stripped_str)

split_stripped_str_list_1 = stripped_str.split(",")
print("Split string:", split_stripped_str_list_1)

split_stripped_str_list_2 = [item.strip() for item in split_stripped_str_list_1]
print("Split string (stripped):", split_stripped_str_list_2)
```

## Classify each list entry

```python
working_list = list.copy(split_stripped_str_list_2)
print("Working list:", working_list)
print()

for item in working_list:
    print(f"Got string item: {item}")
    float_item = float(item)
    print(f"  as float: {float_item}")

    if float_item.is_integer():
        int_item = int(float_item)
        print(f"  as integer: {int_item}")
        print(f"  Checking if {int_item} is prime...")
        if is_prime(int_item):
            print(int_item, " is a prime")
        else:
            print(int_item, " is not a prime")
    else:
        print(item, " is not an integer. Skipping...")
    print()
```

## Takeaways

- Always validate that a numeric token is **integer-valued** before primality tests.
- Wheel factorization (6k±1 style increments after checking 2 and 3) cuts trial work substantially.
