## Overview

Computes the greatest common divisor for all pairs of integers from 1 to 10 using a subtraction-based Euclidean algorithm, with `value` attributes on the function arguments.


## Step-by-step explanation

1. Nested `do` loops walk every pair (a, b) with a and b in 1..MAX_VALUE.
2. Each pair is printed with gcd(a, b) using an explicit integer format.
3. The contained function `gcd` takes `integer, value` arguments so it works on local copies.
4. The loop `do while (a /= b)` implements the subtractive Euclidean algorithm.
5. When a equals b, that common value is the GCD and is returned.
