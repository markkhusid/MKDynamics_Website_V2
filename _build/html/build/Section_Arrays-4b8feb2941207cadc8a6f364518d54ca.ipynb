In this section we follow examples from **Fortran for scientific programming** (FutureLearn MOOC / [gjbex/Fortran-MOOC](https://github.com/gjbex/Fortran-MOOC)). These notebooks demonstrate Fortran arrays: allocation, sections, matrices, elemental procedures, and simple performance experiments, each built and run with **fpm**.
