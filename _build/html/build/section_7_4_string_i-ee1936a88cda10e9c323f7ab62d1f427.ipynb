## Overview

Initializing a `char` array from a literal copies characters (including `'\0'`) into a mutable array.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    char s[] = "Hello, world!";  /* mutable array copy of the literal */

    for (int i = 0; i < 13; i++)
        printf("%c", s[i]);
    printf("\n");

    s[0] = 'J';
    printf("%s\n", s);
}
```


## Step-by-step explanation

1. Initialize `char s[]` from a literal.
2. Print characters, then mutate `s[0]` and print the whole string again.
