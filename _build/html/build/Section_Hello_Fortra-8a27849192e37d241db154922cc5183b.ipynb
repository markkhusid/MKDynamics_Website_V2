## Overview

A minimal Fortran program that prints a greeting. This is the traditional first program when learning a language.


## Step-by-step explanation

1. The program unit is declared with `program hello_fortran` and closed with `end program`.
2. `print *, ...` writes to standard output using list-directed formatting (the `*` format).
3. The character literal is the message that appears on the console.
4. No variables, modules, or command-line arguments are required.
