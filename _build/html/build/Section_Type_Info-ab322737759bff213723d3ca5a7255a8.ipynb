## Overview

Prints kind parameters and numeric inquiry results (huge, tiny, digits, range) for integer and real kinds from iso_fortran_env.


## Step-by-step explanation

1. Kinds INT8/16/32/64 and REAL32/64/128 are imported from iso_fortran_env.
2. For each integer kind, huge, digits, and range describe magnitude and width.
3. For each real kind, huge and tiny show extreme finite magnitudes.
4. Further real inquiries characterize precision and exponent range.
5. This inspects what the compiler actually provides for each kind.
