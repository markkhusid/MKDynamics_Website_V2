## Overview

Pointer variables have types of the form “pointer to T”. The type of the pointed-to object determines how the compiler interprets reads and writes through the pointer.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    int i = 42;
    int *p;          /* type: pointer to int */
    double d = 3.14;
    double *pd;      /* type: pointer to double */

    p = &i;
    pd = &d;

    printf("i = %d,  &i = %p,  p = %p,  *p = %d\n",
           i, (void *)&i, (void *)p, *p);
    printf("d = %f,  &d = %p,  pd = %p, *pd = %f\n",
           d, (void *)&d, (void *)pd, *pd);
    printf("sizeof(int) = %zu, sizeof(int *) = %zu\n",
           sizeof(int), sizeof(int *));
    printf("sizeof(double) = %zu, sizeof(double *) = %zu\n",
           sizeof(double), sizeof(double *));
}
```


## Step-by-step explanation

1. Declare an `int` and a `double`, plus pointers to each type.
2. Assign each pointer with `&` (address-of) of the matching object.
3. Print values, addresses, and pointer values with `%p` (cast to `void *`).
4. Print `sizeof` of the objects and of the pointer types to show they differ.
