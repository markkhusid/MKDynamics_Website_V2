## Overview

`strlen` from `<string.h>` returns the number of characters before the terminating NUL.


## Program Code


```c
#include <stdio.h>
#include <string.h>

int main(void)
{
    char *s = "Hello, world!";

    printf("The string is %zu bytes long (strlen, not counting '\\0').\n",
           strlen(s));
}
```


## Step-by-step explanation

1. Include `<string.h>`.
2. Call `strlen(s)` and print with `%zu`.
