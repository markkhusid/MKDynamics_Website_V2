## Overview

C strings end with a `'\0'` byte. Algorithms walk memory until that terminator.


## Program Code


```c
#include <stdio.h>

/* Manual strlen: count until NUL terminator */
int my_strlen(char *s)
{
    int count = 0;

    while (s[count] != '\0')
        count++;

    return count;
}

int main(void)
{
    char *s = "Hello!";  /* stored as Hello!\0 */

    printf("s = \"%s\"\n", s);
    printf("my_strlen(s) = %d\n", my_strlen(s));
    printf("chars including '\\0' occupy %zu bytes in a literal array sense\n",
           sizeof "Hello!");
}
```


## Step-by-step explanation

1. Implement a simple `my_strlen` that counts until `'\0'`.
2. Compare against a known literal and show `sizeof` of a string literal includes the NUL.
