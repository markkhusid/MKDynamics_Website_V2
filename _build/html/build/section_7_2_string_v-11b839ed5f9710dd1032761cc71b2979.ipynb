## Overview

A common way to hold a string is a `char *` pointing at a string literal.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    char *s = "Hello, world!";

    printf("%s\n", s);
    printf("s points to address %p\n", (void *)s);
}
```


## Step-by-step explanation

1. Assign a literal to `char *s`.
2. Print the string with `%s` and show the pointer address.
