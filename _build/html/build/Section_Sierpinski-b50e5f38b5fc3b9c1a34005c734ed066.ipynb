## Overview

Generates Sierpinski triangle points via the chaos game: repeatedly jump halfway toward a random vertex.


## Step-by-step explanation

1. Read the iteration count from the command line.
2. Initialize three triangle vertices and a random starting point.
3. Each iteration picks a random vertex and moves halfway toward it.
4. Print coordinates for external plotting.
5. After many iterations the attractor is the Sierpinski gasket.
