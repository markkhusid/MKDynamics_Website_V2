## Overview

Reads real numbers from standard input until EOF and prints their sum, skipping malformed lines.


## Step-by-step explanation

1. Initialize total to 0.
2. Repeatedly read a real from input_unit.
3. On end-of-file (stat < 0), exit the loop.
4. On other I/O errors (stat > 0), cycle to skip the bad record.
5. Valid values are accumulated; the final total is printed.
