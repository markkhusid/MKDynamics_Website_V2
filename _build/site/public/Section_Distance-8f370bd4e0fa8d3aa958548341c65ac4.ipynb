## Overview

Reads 2D point pairs (x1,y1,x2,y2) from stdin and prints Euclidean distances.


## Step-by-step explanation

1. Loop reads four reals per record until EOF.
2. I/O errors abort with a message; EOF ends cleanly.
3. distance computes sqrt((x1-x2)^2 + (y1-y2)^2) in double precision.
4. Each distance is printed with list-directed format.
5. Useful pattern for stream processing numeric records.
