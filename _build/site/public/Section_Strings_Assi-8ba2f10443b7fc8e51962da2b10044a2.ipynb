## Overview

Demonstrates fixed-length character variables, assignment truncation/padding, and len vs len_trim.


## Step-by-step explanation

1. A character scalar holds a single character.
2. character(len=5) is a fixed-length string of length 5.
3. Assigning a longer literal truncates to length 5.
4. Assigning a shorter literal pads with blanks; len stays 5 while len_trim drops trailing spaces.
5. Fixed-length strings are a foundation before deferred-length allocatable strings.
