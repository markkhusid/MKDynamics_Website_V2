## Overview

Variant of sum_values that filters non-finite contributions when summing stdin reals.


## Step-by-step explanation

1. Read values from standard input until EOF.
2. Skip or reject non-finite values according to the source checks.
3. Accumulate the remaining finite values.
4. Print the total.
5. Useful when data streams may contain Inf or NaN.
