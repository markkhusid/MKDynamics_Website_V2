## Overview

Monte Carlo estimate of the probability that a needle crosses parallel lines, compared with the analytic Buffon formula involving pi.


## Step-by-step explanation

1. Optional CLI args set the number of needles and the line spacing.
2. Each needle has midpoint x and angle theta (derived type needle_t).
3. Needles are dropped with random position/orientation.
4. is_needle_crossing tests geometric intersection with the nearest line.
5. Empirical probability hits/n is compared to 2*L/(pi*d).
