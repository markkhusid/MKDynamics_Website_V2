## Overview

Reads phrases from standard input and prints whether each is a palindrome.


## Step-by-step explanation

1. Loop reading until end-of-file.
2. is_palindrome compares characters from both ends toward the center.
3. len_trim ignores trailing blanks in the buffer.
4. On the first mismatch return false; if all pairs match return true.
5. Output is a Fortran logical (T or F) per input string.
