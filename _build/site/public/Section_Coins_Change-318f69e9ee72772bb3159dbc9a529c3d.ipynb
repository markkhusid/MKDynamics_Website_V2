## Overview

Computes making change using greedy and dynamic-programming algorithms, then shows coin counts.


## Step-by-step explanation

1. CLI: amount followed by coin denominations (for example 30 1 5 10 25).
2. Greedy repeatedly takes as many of the largest coin as possible.
3. DP builds optimal substructure for the minimum number of coins.
4. Both the number of coins and the multiset used are printed.
5. Compare algorithms on canonical vs non-canonical coin systems.
