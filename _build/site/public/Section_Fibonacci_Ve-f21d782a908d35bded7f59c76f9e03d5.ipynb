## Overview

Compares five Fibonacci algorithms (recursive, skip-ahead, memoized, iterative, closed form) for n = 0..15.


## Step-by-step explanation

1. fib_mod implements multiple Fibonacci strategies.
2. Procedure pointers allow calling each algorithm uniformly.
3. For each n, every algorithm result is printed in a table.
4. Agreement across methods is a sanity check before larger benchmarks.
5. Naive recursion is fine for small n only.
