## Overview

Demonstrates C-interoperable enumerators (enum, bind(C)) for named integer constants.


## Step-by-step explanation

1. iso_c_binding enables C interoperability kinds.
2. enum, bind(C) defines red, green, and blue as sequential enumerator constants.
3. A variable of kind c_int can store enumerator values.
4. Printing red shows the underlying integer code.
5. Assigning color = blue shows that enumerators behave like integer constants.
