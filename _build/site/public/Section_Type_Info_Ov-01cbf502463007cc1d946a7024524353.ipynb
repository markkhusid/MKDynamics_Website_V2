## Overview

Illustrates wrap-around of 8-bit integers when incremented past huge(INT8).


## Step-by-step explanation

1. An integer(kind=INT8) is set near the top of its range (125).
2. A loop adds 1_INT8 six times and prints the value each iteration.
3. Once the value exceeds 127, two's-complement wrap-around yields negatives.
4. Fortran integers do not automatically trap overflow by default.
5. Choosing an appropriate kind (or checking bounds) is required for safe arithmetic.
