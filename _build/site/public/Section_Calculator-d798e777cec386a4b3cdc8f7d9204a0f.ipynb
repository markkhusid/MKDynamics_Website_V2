## Overview

A tiny command-line calculator that parses an expression string `x op y` and evaluates +, -, or *.


## Step-by-step explanation

1. Exactly one command-line argument is expected: a string such as "3.5 + 2.1".
2. `get_expression` uses list-directed read on that string to extract x, op, and y.
3. `calculate` uses select case (op) to choose addition, subtraction, or multiplication.
4. Unsupported operators trigger an error on error_unit and stop.
5. The result is printed in scientific format using double precision (REAL64).
