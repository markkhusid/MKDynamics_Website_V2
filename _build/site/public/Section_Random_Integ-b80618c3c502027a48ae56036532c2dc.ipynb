## Overview

Generates random integers uniformly in an inclusive range [lower, upper].


## Step-by-step explanation

1. CLI requires three integers: count, lower bound, upper bound.
2. random_integer maps a real in [0,1) with floor into the inclusive range.
3. The main loop prints one random integer per line.
4. This is the standard way to get discrete uniforms from random_number.
