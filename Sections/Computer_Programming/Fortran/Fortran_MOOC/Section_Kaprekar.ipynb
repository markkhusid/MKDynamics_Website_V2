## Overview

Applies Kaprekar's process to a 4-digit-like integer until it reaches a fixed point (classically 6174).


## Step-by-step explanation

1. Read a command-line integer in 1..9999 with not all digits identical.
2. Split into digits, sort, form smallest and largest rearrangements.
3. Replace the number by large - small and repeat.
4. Stop when the value no longer changes; print the fixed point.
5. A classic recreational-math algorithm using arrays and sorting.
