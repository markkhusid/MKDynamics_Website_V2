## Overview

Demonstrates a recursive function in Fortran by computing 5!.


## Step-by-step explanation

1. The main program sets n = 5 and prints factorial(n).
2. `factorial` is declared as a recursive function with a result variable.
3. Base case: if n < 2, the result is 1.
4. Recursive case: fac = n * factorial(n-1).
5. This matches the mathematical definition n! = n*(n-1)! with 0! = 1! = 1.
