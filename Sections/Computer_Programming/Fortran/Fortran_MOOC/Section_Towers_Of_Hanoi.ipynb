## Overview

Recursive solution of Towers of Hanoi with explicit stack arrays and move counting.


## Step-by-step explanation

1. Read the number of disks from the command line.
2. init_stacks places all disks on stack 1 in decreasing size.
3. Recursive move_disks moves n-1 aside, moves the base disk, then moves n-1 on top.
4. Each single-disk move updates the array representation and can print stacks.
5. The classic move count is 2^n - 1.
