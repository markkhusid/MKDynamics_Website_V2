## Overview

Reads a year from the command line and reports whether it is a Gregorian leap year.


## Step-by-step explanation

1. `get_year()` requires exactly one command-line argument and reads it as an integer.
2. `is_leapyear(year)` applies the Gregorian rule: divisible by 4, except century years must be divisible by 400.
3. Years not divisible by 4 are never leap years.
4. Century years divisible by 100 but not 400 (for example 1900) are not leap years; 2000 is.
5. The main program prints a human-readable sentence with the result.
