## Overview

Benchmarks x = alpha*x + y in double, mixed, and mixed-with-explicit-conversion modes for a large vector length.


## Step-by-step explanation

1. Allocate and random-fill vectors at the requested length.
2. Time pure double-precision update and print a checksum sum.
3. Time mixed precision (single x with double y/alpha).
4. Time mixed precision with explicit conversion of y to single.
5. Compare timings and checksums to see kind tradeoffs.
