# fplotlib demo in Jupyter

Adapted from [fplotlib Demo](https://fplotlib.fortran-lang.org/index.html)

```fortran
use fplotlib
use lfortran_display
integer, parameter :: n = 200
real(dp) :: x(n), y1(n), y2(n)
integer :: i
real(dp), parameter :: pi = 3.14159265358979323846_dp
```

```fortran
pi
```

```fortran
do i = 1, n
    x(i) = 4.0_dp * pi * real(i - 1, dp) / real(n - 1, dp)
    y1(i) = sin(x(i)) * exp(-0.1_dp * x(i))
    y2(i) = cos(x(i)) * exp(-0.1_dp * x(i))
end do
```

```fortran
print *, x(:5)
```

```fortran
print *, y1(:5)
```

```fortran
print *, y2(:5)
```

```fortran
call clf()
call plot(x, y1, "b-", label="damped sin", lw=2.0_dp)
call title("fplotlib demo")
call xlabel("time")
call ylabel("amplitude")
call grid(.true.)
call legend()
call savefig("demo_damped_sine.svg")
print *, "Wrote demo_damped_sine.svg"
!call display_data("image/svg+xml", render_svg())
```

![Damped sine](demo_damped_sine.svg)

```fortran
call clf()
call plot(x, y1, "b-", label="damped sin", lw=2.0_dp)
call plot(x, y2, "r--", label="damped cos")
call title("fplotlib demo")
call xlabel("time")
call ylabel("amplitude")
call grid(.true.)
call legend()
call savefig("demo_both.svg")
print *, "Wrote demo_both.svg"
!call display_data("image/svg+xml", render_svg())
```

![Both](demo_both.svg)

```fortran
call clf()
call subplot(2, 1, 1)
call plot(x, y1, "b-", label="damped sin", lw=2.0_dp)
call title("top")
call legend()
call grid(.true.)
call constrained_layout(.true.)

call subplot(2, 1, 2)
call plot(x, y2, "r--", label="damped cos")
call title("bottom")
call legend()
call grid(.true.)
call suptitle("Subplots!")
call savefig("demo_subplots.svg")
print *, "Wrote demo_subplots.svg"
!call display_data("image/svg+xml", render_svg())
```
