## Overview

In multi-declarator statements, the `*` binds to the name, not the base type keyword—so `int *p, q` makes only `p` a pointer.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    int a = 1, b = 2;
    int *p, q;       /* p is int*, q is int */
    int *r, *s;      /* both pointers */

    p = &a;
    q = b;           /* not a pointer */
    r = &a;
    s = &b;

    printf("a=%d b=%d q=%d\n", a, b, q);
    printf("*p=%d *r=%d *s=%d\n", *p, *r, *s);
    printf("Note: 'int *p, q' makes only p a pointer.\n");
}
```


## Step-by-step explanation

1. Declare mixed pointer and non-pointer names in one statement.
2. Assign addresses only to the pointer variables.
3. Print values to confirm `q` is not a pointer.
4. Prefer `int *p, *r` (or one declaration per line) for clarity.
