## Overview

Even when declared as `char *`, you can index a string with `s[i]` because indexing is defined in terms of pointer arithmetic.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    char *s = "Hello, world!";   /* char* here */

    for (int i = 0; i < 13; i++)
        printf("%c\n", s[i]);    /* index like an array */
}
```


## Step-by-step explanation

1. Point `s` at a literal.
2. Loop `i` from 0 to 12 and print each character with `%c`.
