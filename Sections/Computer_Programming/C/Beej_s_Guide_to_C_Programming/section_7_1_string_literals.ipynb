## Overview

String literals are sequences of characters in double quotes; `printf` can print them with `%s`.


## Program Code


```c
#include <stdio.h>

int main(void)
{
    printf("%s", "Hello, world!\n");
    printf("%s\n", "This is a test.");
    printf("%s\n", "Quotes inside: \"yes\".");
}
```


## Step-by-step explanation

1. Print several string literals, including escape sequences.
2. Note that literals are typically stored in read-only memory.
