## Overview

C does not store array length with the array. In the declaring scope you can use `sizeof a / sizeof a[0]`, but that fails after the array decays to a pointer in a function parameter.


## Program Code


```c
#include <stdio.h>

void show_length_trick(void)
{
    int x[12];

    printf("sizeof x          = %zu bytes\n", sizeof x);
    printf("sizeof(int)       = %zu bytes\n", sizeof(int));
    printf("elements in x     = %zu\n", sizeof x / sizeof(int));
}

void foo(int x[12])
{
    /* x decays to a pointer here — sizeof no longer means "array size" */
    printf("inside foo: sizeof x = %zu (pointer width, not 12*int)\n", sizeof x);
    printf("inside foo: sizeof x / sizeof(int) = %zu (WRONG as element count)\n",
           sizeof x / sizeof(int));
}

int main(void)
{
    show_length_trick();
    int a[12] = {0};
    foo(a);
}
```


## Step-by-step explanation

1. Compute element count with `sizeof x / sizeof(int)` in the declaring scope.
2. Pass the same array to a function and show `sizeof x` is now a pointer width.
3. Conclude that lengths must usually be passed separately to functions.
